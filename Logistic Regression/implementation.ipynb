```python
def cost(y_true,y_pred):
    y1=y_true*np.log(y_pred)
    y2=(1-y_true)*log(1-y_pred)
    return -np.mean(y1+y2)
```

```python
X=np.random.rand(100,1)
X_b = np.c_[np.ones((100, 1)), X]
Y = (X > 0.5).astype(int)
    
```

```python
plt.scatter(X,Y)
plt.show()
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def gradientdes(X,Y):
    theta=np.random.randn(2,1)
    learning_rate=0.1
    iterations=150
    for interation in range(iterations):
        z=np.dot(X,theta)
        y_pred= sigmoid(z)
        dcost=1/100*(np.dot(X.T,(y_pred - Y)))
        theta=theta-learning_rate*dcost
    return theta
```

```python
theta_f=gradientdes(X_b,Y)
```
