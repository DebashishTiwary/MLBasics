```python
#1st approach decreasing m and b
def gradient_descent(x,y):
    m_curr=b_curr=0
    iterations=1000
    n=len(x)
    learning_rate=0.05
    for i in range(iterations):
        y_pred=b_curr + m_curr*x
        md = -(2/n)*sum(x*(y-y_pred))
        bd = -(2/n)*sum(y-y_pred)

        m_curr=m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd

        cost = (1/n)*sum(val**2 for val in (y-y_pred))
        #print("m {},b {},cost {},iteration {}".format(m_curr,b_curr,cost,i))
    return m_curr,b_curr
```

```python
def linear_regression(x,y):
    x_train=x[0:3]
    y_train=y[0:3]
    x_test=x[3:]
    y_real=y[3:]
    
    m,b=gradient_descent(x_train,y_train)
    y_pred= m*x_test+b
    print("y_predicted {},y_actual {}".format(y_pred,y_real))
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,7,9,11,13])

linear_regression(x,y)
```

```python
X=np.random.rand(100,1)
X_b = np.c_[np.ones((100, 1)), X]
Y= 4+3*X
```

```python
def gradientdes(X,Y):
    theta=np.random.randn(2,1)
    # print(theta)
    learning_rate=0.1
    iterations=150
    for interation in range(iterations):
        y_pred= np.dot(X,theta)
        dcost=2/100*(np.dot(X.T,(y_pred - Y)))
        theta=theta-learning_rate*dcost
    return theta
```
