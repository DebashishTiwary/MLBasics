```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pickle
```

```python
df=pd.read_csv('leetcode_3400_problems_dataset.csv')
```

```python
tags_series = df['Topic_Tags'].fillna('').apply(
    lambda x: [t.strip() for t in x.split(',') if t.strip()]
)
dummies = pd.get_dummies(tags_series.explode()).groupby(level=0).sum()
dummies = (dummies > 0).astype(int)  

df_topic = pd.concat([df, dummies], axis=1)
```

```python
df_topic_diff=pd.get_dummies(df_topic,columns=['Difficulty'],dtype=int)
df_topic_diff_filtercol=df_topic_diff.drop(columns=['Topic_Tags','Problem_Name','Problem_ID'])
```

```python
df_topic_diff_filtercol_med=df_topic_diff_filtercol.drop(columns=['Time_Expert_Mins','Time_Beginner_Mins'])
```

```python
Y=df_topic_diff_filtercol_med['Time_Average_Mins']
X=df_topic_diff_filtercol_med.drop(columns=['Time_Average_Mins'])
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42,shuffle=True)
```

```python
model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
```

```python
model.fit(X_train,Y_train)
```

```python
Y_pred=model.predict(X_test)
```

```python
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)
mae,mse,r2
```

```python
model_filename = "Intermediate_model.pkl"
with open(model_filename, "wb") as file:
    pickle.dump(model, file)
```

```python
# import pandas as pd

# Assuming X_train and X_test are your pandas DataFrames (or you can apply this to test_df)
# training_columns = [...] -> your exact training columns list

def align_test_columns(test_df, training_cols):
    # 1. Create a copy to avoid SettingWithCopyWarning
    df = test_df.copy()
    
    # 2. Map known naming mismatches from test -> train
    rename_mapping = {
        'Arrays': 'Array',
        'BFS': 'Breadth-First Search',
        'DFS': 'Depth-First Search',
        'Divide & Conquer': 'Divide and Conquer',
        'Graphs': 'Graph',
        'Hashing': 'Hash Table',
        'Heap': 'Heap (Priority Queue)',
        'Strings': 'String',
        'Trees': 'Tree'
    }
    df = df.rename(columns=rename_mapping)
    
    # 3. Add any missing training columns as 0 (or False)
    for col in training_cols:
        if col not in df.columns:
            df[col] = 0  # Use False if your data is boolean
            
    # 4. Drop extra columns not present in training and reorder to match training precisely
    df = df[training_cols]
    
    return df

# Example usage:
# X_test_aligned = align_test_consult(X_test, training_columns)
```
