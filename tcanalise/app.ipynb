```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from leetcode_scraper import LeetcodeScraper
```

```python
df=pd.read_csv('leetcode_3400_problems_dataset.csv')
```

```python
df['Topic_Tags'].unique()
```

```python
tags_series = df['Topic_Tags'].fillna('').apply(
    lambda x: [t.strip() for t in x.split(',') if t.strip()]
)
dummies = pd.get_dummies(tags_series.explode()).groupby(level=0).sum()
dummies = (dummies > 0).astype(int)  

df_topic = pd.concat([df, dummies], axis=1)
```

```python
df_topic
```

```python
df_topic_diff=pd.get_dummies(df_topic,columns=['Difficulty'],dtype=int)
```

```python
df_topic_diff
```

```python
df_topic_diff_filtercol=df_topic_diff.drop(columns=['Topic_Tags','Problem_Name','Problem_ID'])
```

```python
df_topic_diff_filtercol
```

```python
df_topic_diff_filtercol_beg=df_topic_diff_filtercol.drop(columns=['Time_Average_Mins','Time_Expert_Mins'])
df_topic_diff_filtercol_med=df_topic_diff_filtercol.drop(columns=['Time_Expert_Mins','Time_Beginner_Mins'])
df_topic_diff_filtercol_exp=df_topic_diff_filtercol.drop(columns=['Time_Beginner_Mins','Time_Average_Mins'])
```

```python
df_topic_diff_filtercol_beg
```

```python
Y=df_topic_diff_filtercol_beg['Time_Beginner_Mins']
```

```python
X=df_topic_diff_filtercol_beg.drop(columns=['Time_Beginner_Mins'])
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42,shuffle=True)
```

```python
model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
```

```python
model.fit(X_train,Y_train)
```

```python
Y_pred=model.predict(X_test)
```

```python
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)
```
