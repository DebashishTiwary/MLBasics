```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('leetcode_3400_problems_dataset.csv')
```

```python
df['Topic_Tags'].head()
```

```python
tags_series = df['Topic_Tags'].fillna('').apply(
    lambda x: [t.strip() for t in x.split(',') if t.strip()]
)
dummies = pd.get_dummies(tags_series.explode()).groupby(level=0).sum()
dummies = (dummies > 0).astype(int)  # Ensure strict binary formatting

df_topic = pd.concat([df, dummies], axis=1)
```

```python
df_topic
```

```python
df_topic_diff=pd.get_dummies(df_topic,columns=['Difficulty'],dtype=int)
```
